Maximize the profit 3x + 2y subject to x + y ≤ 4, x + 3y ≤ 6, x, y ≥ 0. The optimum of a linear program is at a vertex of the feasible polygon.

```python
import calc
calc.eval(r"maximize(3*x+2*y,[x+y<=4,x+3*y<=6,x>=0,y>=0],[x,y])")
```

The same in matrix form: linprog minimizes c·x subject to A·x ≤ b with x ≥ 0 implied, so minimize −3x − 2y:

```python
import calc
calc.eval(r"linprog([-3,-2],[[1,1],[1,3]],[4,6])")
```

Duality. The dual problem, minimize 4u + 6v subject to u + v ≥ 3, u + 3v ≥ 2, u, v ≥ 0, has the same optimal value (strong duality); u and v are the shadow prices of the two constraints.

```python
import calc
calc.eval(r"minimize(4*u+6*v,[u+v>=3,u+3*v>=2,u>=0,v>=0],[u,v])")
```

The two constraint lines y = 4 − x and y = (6 − x)/3; the feasible region is below both, in the first quadrant:

```python
import calc
calc.plot(r"4-x", 0, 6)
```

Special results. No feasible point: the least value of the empty set, inf. Unbounded below: −inf.

```python
import calc
calc.eval(r"linprog(x+y,[x+y>=2,x+y<=1],[x,y])")
```
