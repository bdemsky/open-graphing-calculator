Conventions of this calculator: Laplace F(s) = ∫ f(t)·exp(−st) dt from 0 to ∞; Fourier F(ω) = (1/√(2π))·∫ f(x)·exp(−iωx) dx over the real line (unitary, angular frequency); z-transform X(z) = Σ x(n)·z⁻ⁿ, n ≥ 0.

Laplace: shift in s. exp(−2t)·cos 3t has the transform of cos 3t with s replaced by s + 2:

```python
import calc
calc.eval(r"laplace(exp(-2*t)*cos(3*t))")
```

```python
import calc
calc.eval(r"ilaplace((s+2)/((s+2)^2+9))")
```

Fourier. The Gaussian exp(−x²/2) is its own transform; the box of width 2 gives a sinc:

```python
import calc
calc.eval(r"fourier(exp(-x^2/2),x,w)")
```
