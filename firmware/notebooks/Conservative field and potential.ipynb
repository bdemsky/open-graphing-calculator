F = (2xy + z³, x², 3xz²). On a simply connected region F is conservative (F = ∇f) exactly when ∇×F = 0; then the work ∫F·dr depends only on the end points: f(end) − f(start).

```python
import calc
calc.eval(r"fld:=[2*x*y+z^3,x^2,3*x*z^2]")
```

```python
import calc
calc.eval(r"curl(fld,[x,y,z])")
```

The curl is 0, so a potential exists:

```python
import calc
calc.eval(r"pot:=potential(fld,[x,y,z])")
```

```python
import calc
calc.eval(r"grad(pot,[x,y,z])-fld")
```

Path independence. Work from (0,0,0) to (1,1,1) along the straight line and along the twisted cubic (t, t², t³): both equal f(1,1,1) − f(0,0,0) = 2.

```python
import calc
calc.eval(r"lineint(fld,[x,y,z],[t,t,t],t,0,1)")
```

```python
import calc
calc.eval(r"lineint(fld,[x,y,z],[t,t^2,t^3],t,0,1)")
```

A warning. G = (−y, x)/(x² + y²) has scalar curl 0 but is defined on the plane without the origin, which is not simply connected: its circulation around the unit circle is 2π, so it has no potential there.

```python
import calc
calc.eval(r"curl([-y/(x^2+y^2),x/(x^2+y^2)],[x,y])")
```

```python
import calc
calc.eval(r"lineint([-y/(x^2+y^2),x/(x^2+y^2)],[x,y],[cos(t),sin(t)],t,0,2*pi)")
```
