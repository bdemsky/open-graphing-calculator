Solve the initial value problem y'' + 2y' + 5y = 0, y(0) = 1, y'(0) = 0, then check the answer three ways: substitute it back, check the initial conditions, and solve again by the Laplace transform.

Characteristic equation r² + 2r + 5 = 0: r = −1 ± 2i, so the solutions are exp(−x)·cos 2x and exp(−x)·sin 2x.

The general solution, with the constants c1 and c2:

```python
import calc
calc.eval(r"gsol:=dsolve(y''+2*y'+5*y=0)")
```

The initial conditions go into dsolve as y(0)=1 and y'(0)=0:

```python
import calc
calc.eval(r"odesol:=dsolve([y''+2*y'+5*y=0,y(0)=1,y'(0)=0],y,x)")
```

Substitute. The residual y'' + 2y' + 5y must simplify to 0:

```python
import calc
calc.eval(r"simplify(diff(odesol,x,2)+2*diff(odesol,x)+5*odesol)")
```

Initial conditions [y(0), y'(0)]:

```python
import calc
calc.eval(r"[subst(odesol,x,0),subst(diff(odesol,x),x,0)]")
```

Independence. The Wronskian of the two basis solutions is never 0, so they are independent:

```python
import calc
calc.eval(r"wronskian([exp(-x)*cos(2*x),exp(-x)*sin(2*x)],x)")
```

Laplace transform. L(y'') = s²Y − s y(0) − y'(0) and L(y') = sY − y(0) turn the equation into (s² + 2s + 5) Y = s + 2. Invert Y:

```python
import calc
calc.eval(r"ilaplace((s+2)/(s^2+2*s+5))")
```
