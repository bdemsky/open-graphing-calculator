A is diagonalizable when it has n independent eigenvectors; with them as the columns of P, A = P·D·P⁻¹ and D is diagonal. Then powers and exponentials are easy: Aⁿ = P·Dⁿ·P⁻¹, exp(At) = P·exp(Dt)·P⁻¹.

```python
import calc
calc.eval(r"mat:=[[4,1],[2,3]]")
```

```python
import calc
calc.eval(r"eigenvects(mat)")
```

Eigenvalue 2 has eigenvector (−1/2, 1), so also (−1, 2); eigenvalue 5 has (1, 1). Put them in the columns of P:

```python
import calc
calc.eval(r"pmat:=[[-1,1],[2,1]]")
```

```python
import calc
calc.eval(r"inv(pmat)*mat*pmat")
```

Powers. A¹⁰ = P·D¹⁰·P⁻¹ (the difference is the zero matrix):

```python
import calc
calc.eval(r"pmat*diag([2^10,5^10])*inv(pmat)-mat^10")
```

Exponential, for the system x' = A·x, x(t) = exp(At)·x(0):

```python
import calc
calc.eval(r"expm(mat*t)")
```

Symmetric matrices have orthonormal eigenvectors: A = Q·D·Q^T with Q orthogonal (the spectral theorem).

```python
import calc
calc.eval(r"eigenvects([[2,1],[1,2]])")
```

```python
import calc
calc.eval(r"transpose([[1,1],[-1,1]]/sqrt(2))*[[2,1],[1,2]]*([[1,1],[-1,1]]/sqrt(2))")
```

A matrix with a repeated eigenvalue but too few eigenvectors is not diagonalizable; its Jordan form has a 1 above the diagonal:

```python
import calc
calc.eval(r"eigenvects([[1,1],[0,1]])")
```

```python
import calc
calc.eval(r"jordan([[2,1],[-1,0]])")
```
