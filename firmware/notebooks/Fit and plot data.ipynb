Least squares: minimize ‖A·c − y‖ over the coefficients c. The solution is c = A⁺·y (pseudoinverse), the same as the normal equations A^T A·c = A^T y when the columns of A are independent.

```python
import calc
calc.eval(r"xs:=[1,2,3,4,5]")
```

```python
import calc
calc.eval(r"ys:=[2.2,4.1,6.2,7.9,10.1]")
```

A line y = c₀ + c₁x. The built-in regression:

```python
import calc
calc.eval(r"linreg(xs,ys)")
```

The same from the design matrix with columns 1 and x:

```python
import calc
calc.eval(r"pinv(transpose([[1,1,1,1,1],xs]))*ys")
```

A parabola y = c₀ + c₁x + c₂x²: add a column x².

```python
import calc
calc.eval(r"pinv(transpose([[1,1,1,1,1],xs,xs^2]))*ys")
```

The fitted line and parabola (almost the same here):

```python
import calc
calc.plot(r"1.96*x+0.22", 0, 6)
```

Interpolation instead of fitting: the natural cubic spline passes through every point, with continuous slope and curvature.

```python
import calc
calc.eval(r"spline([0,1,2,3],[0,1,0,1],x)")
```

Each piece is a cubic; at x = 1 and x = 2 the neighbouring pieces agree in value, slope and curvature, and the curvature is 0 at both ends (a natural spline).
